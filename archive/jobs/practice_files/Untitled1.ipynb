```python
from pyspark.sql import SparkSession

# Initialize Spark Session pointing to your standalone Master
spark = SparkSession.builder \
    .appName("Iceberg-Architect-Playground") \
    .master("spark://localhost:7077") \
    .config("spark.jars.packages", "org.apache.iceberg:iceberg-spark-runtime-3.5_2.12:1.4.3") \
    .config("spark.sql.extensions", "org.apache.iceberg.spark.extensions.IcebergSparkSessionExtensions") \
    .config("spark.sql.catalog.local", "org.apache.iceberg.spark.SparkCatalog") \
    .config("spark.sql.catalog.local.type", "hadoop") \
    .config("spark.sql.catalog.local.warehouse", "/opt/spark/warehouse") \
    .getOrCreate()

print("Connected to Master. Iceberg Catalog 'local' is active.")
```
